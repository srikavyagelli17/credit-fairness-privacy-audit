```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, Markdown, display

ROOT = Path.cwd()
if not (ROOT / "results").exists() and (ROOT.parent / "results").exists():
    ROOT = ROOT.parent
TABLES = ROOT / "results" / "tables"
FIGURES = ROOT / "results" / "figures"
pd.set_option("display.max_columns", 40)
pd.set_option("display.width", 120)
```

```python
summary = pd.read_csv(TABLES / "dataset_summary.csv")
performance = pd.read_csv(TABLES / "baseline_performance.csv")
fairness = pd.read_csv(TABLES / "fairness_baseline.csv")
proxy = pd.read_csv(TABLES / "proxy_analysis.csv")
tradeoff = pd.read_csv(TABLES / "reweighing_tradeoff.csv")
privacy = pd.read_csv(TABLES / "dp_sweep_summary.csv")
recommended = pd.read_csv(TABLES / "recommended_config.csv")
display(summary)
display(performance)
display(fairness[fairness["metric"].isin(["dp_difference", "dp_ratio", "eo_difference", "four_fifths_pass"])])
display(proxy[proxy["record_type"].isin(["auc", "point_biserial", "cramers_v"])].sort_values("abs_value", ascending=False).groupby("task").head(8))
display(tradeoff)
display(recommended)
```
